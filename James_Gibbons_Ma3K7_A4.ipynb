```python
import numpy as np

n = int(input("Enter n"))   #taking input for n
A = np.full((n, n), -1)   #creating matrix

for i in range(n*n):   #repeats for every cell
    if i % 2 == 0:   #Player 1 turn
        player = 1
        number = 0
    else:
        player = 2   #Player 2 turn
        number = 1
    print(A)
    print("Player", player, "enter row and column:")
    row = int(input()) - 1   #allows users to enter their number in row and column of choice
    column = int(input()) - 1
    A[row, column] = number   #enters players number into the correct cell

print(A)
det = np.linalg.det(A)   #calculates determinant

print("Determinant is", det)

if round(det) == 0:   #decides who winner is
    print("Player 1 wins!")
else:
    print("Player 2 wins!")
```
